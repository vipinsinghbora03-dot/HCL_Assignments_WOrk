```python
order_id = int(input("Enter the order ID: "))
product = input("Enter the product name: ")
quantity = int(input("Enter the quantity: "))
cancelled = input("Is the order cancelled? (yes/no): ").strip().lower() == 'yes'
if product == "Wireless Mouse":
    unit_price = 399.5
elif product == "Basmati Rice 5kg":
    unit_price = 81.25
elif product == "Denim Jeans":
    unit_price = 228.4285142857
elif product == "Notebook Pack":
    unit_price = 120.0
elif product == "Bluetooth Speaker":
    unit_price = 499.8
elif product == "USB-C Cable":
    unit_price = 299.0
else:
    print("Product is not available in store.")
print("\n------BILL------")
print(f"Order ID: {order_id}")
print(f"Product: {product}")
print(f"Quantity: {quantity}")
if cancelled:
    print("Order Status: Cancelled")
else:
    total_price = unit_price * quantity
    print(f"Total Price: {total_price:.2f}")   
if total_price >= 5000:
    print("Discount: 5%")
    total_price *= 0.95
    print(f"Total Price after discount: {total_price:.2f}")
elif total_price >=10000:
    print("Discount: 10%")
    total_price *= 0.9
    print(f"Total Price after discound: {total_price:.2f}")


```

**T4**

```python
order = [("Wireless Mouse", 2, 799),("Basmati rice 5kg", 8, 650),("Denim Jeans", 7, 1599),("Bluetooth Speaker", 5, 2499),("Notebook pack", 10,120),("USB-C Cable", 0, 299)]
grand_total= 0
for i, order in enumerate(order, start=1):
    item, price, qty = order

    # Skip invalid orders
    if price <=0 or qty <= 0:
        print(f"order {i}: Invalid order skipped")
        continue

    subtotal = price*qty
    discount = subtotal * 0.05      # 5% discount
    gst = (subtotal - discount) * 0.18   # 18% GST
    final = subtotal - discount + gst

    grand_total += final

    print(f"Order {i}: {item}")
    print(f"Subtotal = ₹{subtotal:.2f}")
    print(f"Discount = ₹{discount:.2f}")
    print(f"GST = ₹{gst:.2f}")
    print(f"Final Amount = ₹{final:.2f}")
    print("-" * 30)

print(f"Grand Total = ₹{grand_total:.2f}")



```

```python
attempts = 0

while attempts < 3:
    quantity = int(input("Enter the quantity: "))

    if quantity > 0:
        print("Valid Quantity: ",quantity)
        break
    attempts += 1
    print("Invalit inpuy,attempt left: ", 3 - attempts)

if attempts == 3 and quantity <= 0:
    print("maximum attempts reached.")
```
