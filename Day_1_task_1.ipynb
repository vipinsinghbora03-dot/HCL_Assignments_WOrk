```python
order_id = int(input("Enter the order ID: "))
product = input("Enter the product name: ")
quantity = int(input("Enter the quantity: "))
cancelled = input("Is the order cancelled? (yes/no): ").strip().lower() == 'yes'
if product == "Wireless Mouse":
    unit_price = 399.5
elif product == "Basmati Rice 5kg":
    unit_price = 81.25
elif product == "Denim Jeans":
    unit_price = 228.4285142857
elif product == "Notebook Pack":
    unit_price = 120.0
elif product == "Bluetooth Speaker":
    unit_price = 499.8
elif product == "USB-C Cable":
    unit_price = 299.0
else:
    print("Product is not available in store.")
print("\n------BILL------")
print(f"Order ID: {order_id}")
print(f"Product: {product}")
print(f"Quantity: {quantity}")
if cancelled:
    print("Order Status: Cancelled")
else:
    total_price = unit_price * quantity
    print(f"Total Price: {total_price:.2f}")   
if total_price >= 5000:
    print("Discount: 10%")
    total_price *= 0.9
    print(f"Total Price after discount: {total_price:.2f}")
```
