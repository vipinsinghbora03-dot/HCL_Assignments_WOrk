**Two variables reffering to the same object**

```python
a =[10, 20 , 30]
b = a
print("ID of a:", id(a))
print("ID of b:", id(b))
print("Same object", a is b)  # Check if both variables point to the same object
```

```python
print(10//3)  # Floor division
print(10**3)  # Exponentiation
print(10/3)  # Standard division
```

```python
x = 10 
print("Initial:", x)
x += 5  # Increment by 5
print("After incrementing by 5:", x)
x-= 3  # Decrement by 3
print("After decrementing by 3:", x)
x *= 2  # Multiply by 2
print("After multiplying by 2:", x)
x/=4  # Divide by 4
print("After dividing by 4:", x)
x//=3  # Floor division by 3
print("After floor division by 3:", x)

```

```python
age = 25
has_id= True
print(age >= 18 and has_id)  # Check if age is at least 18 and has ID
print(age < 18 or has_id)  # Check if age is less than 18 or has ID
print(not has_id)  # Negation of has_id
print(True and False)  # Logical AND
print(True or False)  # Logical OR
print(not True)  # Negation of True
```

**Membership Operators**

```python
language = ["Python", "Java", "JQL"]
print("Python" in language)  # Check if "Python" is in the list
print("C++" in language)  # Check if "C++" is in the list
print("C++" not in language)  # Check if "C++" is not in the list
text= "Data Analytics"
print("Data" in text)  # Check if "Data" is in the string
```

```python
list_a = [1, 2, 3]
list_b = list_a  # list_b references the same list as list_a
list_c = [1, 2, 3]  # list_c is a separate list with the same elements
print(list_a is list_b)  # True, same object
print(list_a is list_c)  # False, different objects
print(list_a is not list_c)  # True, list_a is not the same object as list_c
```
