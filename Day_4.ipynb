```python
import numpy as np
arr = np.array([[10, 20, 30], [40, 50, 60]]) # here roe1 = 10(c1), 20(c2), 30(c3) and row2 = 40, 50, 60
print(arr.ndim) # number of dimenssions
print(arr.shape)# rows, columns
print(arr.size)# Total elements
print(arr[0:2,1:3])# Sub-array
```

```python
a = np.array([10, 20, 30, 40,50])

print("Original array:")
print(a)

print("\nAddition:")
print(a + 5)

print(a.ndim)
print(a.shape)# give output as tupel.
print(a.size)
```

```python
marks = np.array([70, 80, 65, 90, 85])
average = np.mean(marks)
print("Average Marks:", average)
```

**Reshaping The Array**

```python
number = np.arange(1, 13)

print("original array:", number)

reshaped = number.reshape(3, 4)
print("\nReshaped array:", reshaped)
print(reshaped.shape)
print(reshaped.size)
print(reshaped.ndim)
```
