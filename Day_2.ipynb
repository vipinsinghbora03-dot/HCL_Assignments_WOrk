**Variable-length argument**

```python
# * for required(Positional) assess an tuple
# ** Keywords
def add(*num):# single star for non-keyword variable length arguments.
    sum = 0
    for n in num:
        sum +=n;
    return sum
print(add())# 0 argument
print(add(1))# 1 argument
print(add(1,2))# 2 arguments
print(add(1,2,3))# 3 arguments
print(add(1,2,3,4))# 4 arguments
```

```python
def display_emp(**data):
    for k,v in data.items():
        print("{}\t:{}".format(k, v))
#calling **Keyword fxn
display_emp(id=1001,name="Thanos")
```

```python
def square_cube(n):
    square= n*n
    cube = square*n
    return square, cube
s,c = square_cube(5) # Here, s and c are converted into tuple.
print("Square of 5 = ",s)
print("cube of 5 = ", c)
```
