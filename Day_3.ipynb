```python
class cat:
    def make_sound(self):
        return "Meow"

class Cow:
    def make_sound(self):
        return "Moo"

class Duck:
    def make_sound(self):
        return "Quack"

animals = [cat(), Cow(), Duck()]
for animal in animals:
    print(animal.make_sound())
```

```python
class SmartPhone:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
        self.battery = 100

    def use_app(self, minutes):
        self.battery -= minutes*0.5
        if self.battery<0:
            self.battery = 0

    def charge(self):
        self.battery = 100

    def status(self):
        print(f"Brand : {self.brand}")
        print(f"Battery: {self.battery}")
        print(f"Model: {self.model}") 

phone = SmartPhone("Realme","RMX 5060")
phone.status()
phone.use_app(50)
phone.status()
phone.charge()
phone.status()


    
    
```

```python
student = {
    "name": "Vipin",
    "age": 21,
    "course": "Data Analytics",
    "marks": 88
}
# Access values using keys.
print("Name:", student["name"])
print("Marks:", student["marks"])
# Safely access a key that may not exist.
print("City:", student.get("city", "Not available"))
# Update an exiting values.
student["marks"]= 92
# add a new value pair.
student["city"]= "Delhi"
# updating the dictionary.
student.update({"age": 22, "status": "Active"})
# Printing the updated dictinory.
print(student)
```
